```python
import sys
from pathlib import Path

project_root = Path.cwd().parent
sys.path.append(str(project_root))

from src.forecasting.feature_engineering import create_features

X, y = create_features(energy_data)

print("Feature matrix:", X.shape)
print("Target:", y.shape)

display(X.head())
display(y.head())
```

```python
# Chronological 80/20 split
split_index = int(len(X) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

print("\nTraining period:")
print(X_train.index.min(), "to", X_train.index.max())

print("\nTesting period:")
print(X_test.index.min(), "to", X_test.index.max())
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Initialize model
model = RandomForestRegressor(
    n_estimators=100,
    max_depth=15,
    min_samples_split=5,
    random_state=42,
    n_jobs=-1
)

# Train
model.fit(X_train, y_train)

# Predict
y_pred = model.predict(X_test)

# Evaluate
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("Random Forest Performance")
print("-------------------------")
print(f"MAE:  {mae:,.2f} MW")
print(f"RMSE: {rmse:,.2f} MW")
print(f"R²:   {r2:.4f}")
```

```python
# Baseline 1: Predict next hour using current demand
baseline_1 = X_test["Demand_Lag_1"]

# Baseline 2: Predict using demand from the same hour yesterday
baseline_24 = X_test["Demand_Lag_24"]

def evaluate_model(name, actual, predicted):
    mae = mean_absolute_error(actual, predicted)
    rmse = np.sqrt(mean_squared_error(actual, predicted))
    r2 = r2_score(actual, predicted)

    print(f"\n{name}")
    print(f"MAE:  {mae:,.2f} MW")
    print(f"RMSE: {rmse:,.2f} MW")
    print(f"R²:   {r2:.4f}")


evaluate_model("Random Forest", y_test, y_pred)
evaluate_model("Persistence Baseline", y_test, baseline_1)
evaluate_model("24-Hour Baseline", y_test, baseline_24)
```

```python
import pandas as pd
import joblib

from pathlib import Path
from src.forecasting.feature_engineering import create_features


# --------------------------------------------------
# 1. Locate project directory
# --------------------------------------------------

project_root = Path.cwd().parent

model_path = (
    project_root
    / "models"
    / "trained"
    / "random_forest.joblib"
)


# --------------------------------------------------
# 2. Load trained Random Forest model
# --------------------------------------------------

model = joblib.load(model_path)

print("Forecasting model loaded successfully!")


# --------------------------------------------------
# 3. Generate forecasting features
# --------------------------------------------------

X, y = create_features(energy_data)


# --------------------------------------------------
# 4. Generate next-hour demand predictions
# --------------------------------------------------

forecast_predictions = model.predict(X)


# --------------------------------------------------
# 5. Align predictions with actual timestamps
# --------------------------------------------------

forecast_results = pd.DataFrame({
    "Timestamp": energy_data["Timestamp"].iloc[24:].values,
    "Forecast_Demand": forecast_predictions,
    "Actual_Demand": y.values
})


# --------------------------------------------------
# 6. Display results
# --------------------------------------------------

print("\nForecasting completed!")
print("Total forecasts generated:", len(forecast_results))

display(forecast_results.head(10))


# --------------------------------------------------
# 7. Calculate forecast performance
# --------------------------------------------------

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae = mean_absolute_error(
    forecast_results["Actual_Demand"],
    forecast_results["Forecast_Demand"]
)

rmse = np.sqrt(mean_squared_error(
    forecast_results["Actual_Demand"],
    forecast_results["Forecast_Demand"]
))

r2 = r2_score(
    forecast_results["Actual_Demand"],
    forecast_results["Forecast_Demand"]
)

print("\nForecast Performance")
print("--------------------")
print(f"MAE:  {mae:,.2f} MW")
print(f"RMSE: {rmse:,.2f} MW")
print(f"R²:   {r2:.4f}")
```
